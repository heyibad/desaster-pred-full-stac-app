## 2  Extract Dataset

```python
from pathlib import Path
import zipfile

zip_name = "dataset.zip" 
extract_dir = Path("disaster_dataset")

# Most Roboflow zips contain train/ and val/ at the top level
data_dir = extract_dir
print(f"Dataset root: {data_dir}")
```

## 4. Train the Model


```python
from ultralytics import YOLO

if data_dir is None:
    raise ValueError("Dataset root not found. Ensure the zip contains train/ and val/ folders.")

model = YOLO("yolov8m-cls.pt")

results = model.train(
    data=str(data_dir),
    epochs=30,
    imgsz=224,
    batch=32,
    device=0,
    patience=10,
    project="disaster_classification",
    name="yolov8m_run"
  )
```

## 5. Evaluate and Validate

```python
# Validate the model
metrics = model.val()
print(metrics.top1) # top1 accuracy
print(metrics.top5) # top5 accuracy
```

## 6. Export and Save the Best Model

```python
import shutil
from pathlib import Path

best_weights_path = Path(results.save_dir) / "weights" / "best.pt"
exported_model_name = Path("best_disaster_classifier.pt")

if best_weights_path.exists():
    shutil.copy(best_weights_path, exported_model_name)
    print(f"Model saved as {exported_model_name}. Download it from the Files panel.")
else:
    print(f"Best weights not found at: {best_weights_path}")
```
